1. What is Version Control?
Version control is a system used to track and manage changes made to files over time.
For example, while developing a Data Science project, you may create several versions of a Python file:
•	sales_analysis.py
•	sales_analysis_v2.py
•	sales_analysis_final.py
•	sales_analysis_final_updated.py
Instead of creating many files manually, version control helps you track changes and return to an earlier version when necessary.


Uses of version control:
•	Track changes in files.
•	Restore previous versions.
•	Collaborate with other developers.
•	Maintain project history.


2. What is Git?
Git is a free, open-source, distributed version control system used to track changes in files.
It was created by Linus Torvalds in 2005.
Example: A Data Scientist modifies a Python file to improve a machine learning model. Git can track the changes and help identify what was modified and when.


Uses of Git:
•	Track changes in source code.
•	Save project versions.
•	Create separate branches for new features.
•	Merge changes from different team members.
•	Work on projects without requiring a constant internet connection.


3. What is GitHub?
GitHub is an online platform used to host Git repositories and collaborate on projects.
It allows developers and Data Scientists to store, share, review, and manage their project files online.
Example: A student completes a customer segmentation project using Python and uploads the code to GitHub to share it with a trainer or potential employer.


Uses of GitHub:
•	Store projects online.
•	Share code with others.
•	Collaborate with team members.
•	Maintain project documentation.
•	Build a portfolio for job applications.
Website: www.github.com

5. Important Git and GitHub terminology
Repository (Repo)
A location that stores project files and their version history.
Example: A repository containing a Python sales prediction project.
Working Directory
The folder where you create or modify your project files.
Staging Area
The intermediate area where you select changes to include in the next commit.
Commit
A saved snapshot of selected changes, accompanied by a message describing them.
Branch
A separate line of development that allows you to work on a feature without directly changing the main branch.
Push
Sends local commits to a remote repository, such as one hosted on GitHub.
Pull
Fetches changes from a remote repository and integrates them into your current local branch.
Clone
Creates a local copy of an existing remote repository.


6. How Git works
The basic Git workflow has three important stages.
1. Working Directory
Create or modify Python files
2. Staging Area
Select changes using git add
3. Local Repository
Save a snapshot using git commit
4. Remote Repository (GitHub)
Upload commits using git push


7. What is Git Bash?
Git Bash is a command-line application included with Git for Windows. It allows users to execute Git commands and Unix-style commands on Windows.
It is useful for students who want to manage projects using commands rather than graphical interfaces.
Example commands:
 git --version
git status
git add .
git commit -m "Initial commit"
To install Git Bash, download Git for Windows from the official Git website.
