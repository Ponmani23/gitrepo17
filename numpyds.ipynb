2. NumPy Array
An array is a collection of elements stored in a structured format. NumPy arrays are especially useful for numerical calculations.

Python lists can store different data types, while NumPy arrays generally use a common data type for their elements.

```python
import numpy as np
marks = np.array([80, 90, 75, 85, 95])
print(marks)
print(type(marks)) #ndarray means N-dimensional array.
```

```python
# since it stores all values in common data type, it will store it as string
stud = np.array([80, 90, 'Anu', 85, 95])
print(stud)
print(type(stud))
```

```python
# Python list, Multiplying a Python list by 2 repeats the list,
#whereas multiplying a NumPy array by 2 multiplies each element by 2.
a = [10, 20, 30]
print(a * 2)
```

```python
# NumPy array- can assign datatype so it memory space can be used efficiently
b = np.array([10, 20, 30], dtype=np.int8)
print(b * 2)
print(b.dtype)
```

```python
#7. Array Slicing -Slicing extracts a portion of an array.
#Syntax: array[start:stop:step]
#The stop index is excluded.
a = np.array([10, 20, 30, 40, 50, 60])
print(a[1:4])
print(a[:3])
print(a[::2])     # slicing a 1dimensional array
print(a[::-1])
```

```python
#12. Comparison and Boolean Operations
#NumPy can compare each array element and return True or False.
a = np.array([10, 20, 30, 40, 50])
print(a > 25)
print(a == 30)
print(a!=50)
```

```python
#Logical operations
#& means element-wise AND.
#| means element-wise OR.
#Use parentheses around individual conditions.
a = np.array([10, 20, 30, 40, 50])
print((a >= 20) & (a <= 40))
print((a < 20) | (a > 40))
```

17. Random Number Generation
Random numbers are useful for simulations, sampling and creating practice datasets.

Modern NumPy code generally uses np.random.default_rng().
i.Random decimal values
ii.Random integers
iii.Normal distribution
iv.Random choice

```python
#Creates five random decimal values from 0 (inclusive) to 1 (exclusive).
rng = np.random.default_rng(42)
print(rng.random(10))
```

```python
#Creates five random integers from 1 to 100.
rng = np.random.default_rng(42)
print(rng.integers(1, 101, size=5))
```

```python
#Generates five values from a normal distribution with mean 50 and
#standard deviation 10.
rng = np.random.default_rng(42)
print(rng.normal(50, 10, size=5))
```

```python
#Selects three values randomly from the given choices.
#The seed 42 makes results reproducible for the same generator and sequence of operations.
rng = np.random.default_rng(42)
print(rng.choice([10, 20, 30, 40], size=3))
```

```python
#18. Handling Missing and Invalid Numerical Values
#In Data Science, datasets can contain missing values or invalid numerical results.
#np.nan - Represents a missing numerical value.
#np.nanmean() calculates the mean while ignoring NaN values.
a = np.array([10.0, 20.0, np.nan, 40.0])
print(np.isnan(a))
print(np.nanmean(a))
```

```python
#np.nan_to_num()-Replaces NaN and, by default, infinite values with finite numbers.
#Replacing missing values with zero is not always appropriate. The correct treatment 
#depends on what the missing data represents.
a = np.array([10.0, np.nan, 30.0])
print(np.nan_to_num(a, nan=0))
```

```python
#19. Copy vs View
#This is an important concept when manipulating datasets.
#Copy -A copy creates an independent array.
a = np.array([10, 20, 30])
b = a.copy()
b[0] = 100
print(a)
print(b)
```

```python
#View - A view shares the underlying data with the original array when possible.
a = np.array([10, 20, 30])
b = a.view()
b[0] = 100
print(a)
```

```python
#Changing a view can affect the original array. Basic slicing usually produces a view, 
#while advanced indexing usually produces a copy.
#20. Matrix Operations and Linear Algebra
#NumPy provides functions for working with matrices, which are important in statistics 
#and machine learning.
#Matrix addition
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print(a + b)
```

```python
#Matrix multiplication
print(a @ b)
```

```python
#Determinant
print(np.linalg.det(a))
```

```python
#Inverse- The inverse exists only for a square matrix with a nonzero determinant,
#apart from numerical precision considerations.
print(np.linalg.inv(a))
```

```python
#Solving linear equations - This solves a system of equations of the form \(Ax=B\).
A = np.array([[2, 1], [1, 3]])
B = np.array([5, 6])
print(np.linalg.solve(A, B))
```
